#Start Visualization 1

Q)How are popularity scores distributed across Spotify tracks?

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="popularity",
    bins=30,
    kde=True
)

plt.title("Distribution of Spotify Track Popularity")
plt.xlabel("Popularity Score")
plt.ylabel("Number of Tracks")

plt.tight_layout()
plt.show()
```

Do songs with higher danceability tend to have higher popularity?

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="danceability",
    y="popularity",
    alpha=0.4
)

plt.title("Danceability vs Track Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity Score")

plt.tight_layout()
plt.show()
```

Question: Which Spotify audio features are most strongly related to each other?

```python
audio_features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
    "popularity"
]

correlation = df[audio_features].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Spotify Audio Features")
plt.tight_layout()
plt.show()
```

Question: How does song duration vary across different music genres?

```python
top_8_genres = df["track_genre"].value_counts().head(8).index
top_8_genres
duration_data = df[df["track_genre"].isin(top_8_genres)]
duration_data
```

```python
# Select the 8 most common genres
top_8_genres = df["track_genre"].value_counts().head(8).index

# Filter the dataset for those genres
duration_data = df[df["track_genre"].isin(top_8_genres)]

# Create the box plot
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=duration_data,
    x="track_genre",
    y="duration_min"
)
plt.title("Song Duration Distribution Across Top 8 Genres")
plt.xlabel("Music Genre")
plt.ylabel("Duration (Minutes)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Question: Do explicit songs have different popularity scores compared with non-explicit songs?

```python
# Create the box plot
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="explicit",
    y="popularity"
)

# Add title and labels
plt.title("Popularity Comparison: Explicit vs Non-Explicit Tracks")
plt.xlabel("Explicit Content")
plt.ylabel("Popularity Score")

# Display the chart
plt.tight_layout()
plt.show()
```

Question: What are the average popularity and duration of Spotify tracks, and which genres have the highest average popularity?

Question: Is there a relationship between danceability and track popularity?

```python
# Calculate correlation
danceability_correlation = df["danceability"].corr(df["popularity"])

# Display the result
print("Correlation between Danceability and Popularity:",
      round(danceability_correlation, 3))
```

Question: Which music genres have the highest and lowest average popularity?

```python
# Calculate average popularity for each genre
genre_popularity = df.groupby("track_genre")["popularity"].mean()

# Identify the most popular genre
most_popular_genre = genre_popularity.idxmax()
highest_popularity = genre_popularity.max()

# Identify the least popular genre
least_popular_genre = genre_popularity.idxmin()
lowest_popularity = genre_popularity.min()

# Display the results
print("Most Popular Genre:", most_popular_genre)
print("Highest Average Popularity:", round(highest_popularity, 2))

print("\nLeast Popular Genre:", least_popular_genre)
print("Lowest Average Popularity:", round(lowest_popularity, 2))
```

Question: Which 10 Spotify tracks have the highest popularity scores?

```python
# Select the top 10 tracks
top_10_tracks = df.nlargest(10, "popularity")

# Create the bar chart
plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_10_tracks,
    x="popularity",
    y="track_name"
)

plt.title("Top 10 Most Popular Spotify Tracks")
plt.xlabel("Popularity Score")
plt.ylabel("Track Name")

plt.tight_layout()
plt.show()
```

Question: Do explicit and non-explicit tracks have different average popularity?

```python
# Calculate average popularity for each group
explicit_avg = df.groupby("explicit")["popularity"].mean().reset_index()

# Create the bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    data=explicit_avg,
    x="explicit",
    y="popularity"
)

plt.title("Average Popularity: Explicit vs Non-Explicit Tracks")
plt.xlabel("Explicit Content")
plt.ylabel("Average Popularity")

plt.tight_layout()
plt.show()
```

Question: Which music genres have the highest and lowest average popularity?

```python
# Calculate average popularity by genre
genre_popularity = df.groupby("track_genre")["popularity"].mean()

# Select the top 15 genres
top_15_genres = genre_popularity.sort_values(
    ascending=False
).head(15)

# Create the chart
plt.figure(figsize=(12, 7))

sns.barplot(
    x=top_15_genres.values,
    y=top_15_genres.index
)

plt.title("Top 15 Music Genres by Average Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Music Genre")

plt.tight_layout()
plt.show()
```

Question: Do energetic songs tend to have higher popularity?

```python
# Create the scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="energy",
    y="popularity",
    alpha=0.4
)

plt.title("Energy vs Track Popularity")
plt.xlabel("Energy")
plt.ylabel("Popularity Score")

plt.tight_layout()
plt.show()
```

Question: How does average popularity change across different energy levels?

```python
# Divide energy into 5 categories
df["energy_level"] = pd.cut(
    df["energy"],
    bins=5,
    labels=["Very Low", "Low", "Medium", "High", "Very High"]
)

# Calculate average popularity for each energy level
energy_popularity = df.groupby(
    "energy_level", observed=False
)["popularity"].mean().reset_index()

# Create the bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=energy_popularity,
    x="energy_level",
    y="popularity"
)

plt.title("Average Popularity by Energy Level")
plt.xlabel("Energy Level")
plt.ylabel("Average Popularity")

plt.tight_layout()
plt.show()
```

Question: Do tracks with different tempos have different average popularity?

```python
# Divide tempo into 5 categories
df["tempo_category"] = pd.cut(
    df["tempo"],
    bins=5,
    labels=["Very Slow", "Slow", "Medium", "Fast", "Very Fast"]
)

# Calculate average popularity
tempo_popularity = df.groupby(
    "tempo_category", observed=False
)["popularity"].mean().reset_index()

# Create the bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=tempo_popularity,
    x="tempo_category",
    y="popularity"
)

plt.title("Average Popularity by Tempo Category")
plt.xlabel("Tempo Category")
plt.ylabel("Average Popularity")

plt.tight_layout()
plt.show()
```
