```python
import pandas as pd
```

```python
df = pd.read_csv('/content/e_commerce_orders_dataset.csv',encoding='latin1')
df
```

To check the columns

```python
df.columns
```

Step 1: Checking the column names

STEP 2 — Check duplicate values

```python
df[df.duplicated]
int(df.duplicated().sum())
```

step3:Remove or Drop the duplicate values

```python
df.drop_duplicates(inplace=True)
df
```

```python
df[df.duplicated()]
int(df.duplicated().sum())
```

STEP 4 — Clean unwanted spaces and capitalization

```python
df['customer_name'] = df['customer_name'].str.strip().str.title()
```

```python
df
```

Step 5: Cleaning category values

```python
df['category'] = df['category'].str.strip().str.title()
df
```

Step 6 — Clean status

```python
df['status'] = df['status'].str.strip().str.lower()
df
```

Step 6: Changing the datatype of order_date

```python
df.info()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'],format='mixed')
df
```

Step 8 — Check NA values

```python
df.isna().sum()
```

Step 9 — Fill missing email and phone

```python
df['email'] = df['email'].fillna('Not Available')
df['phone'] = df['phone'].fillna('Not Available')
df['order_date'] = df['order_date'].fillna(0)
```

```python
df.isna().sum()
```

Step 10 — Fill missing total amount

```python
df['total_amount'] = df['total_amount'].fillna(
    df['quantity'] * df['unit_price']
)
```

Step 11 — Final check

```python
df.info()
```
