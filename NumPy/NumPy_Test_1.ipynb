25.Generate a 5 × 5 matrix of random integers between 1 and 100. Find:
1.Maximum value
2.Minimum value
3.Mean
4.Standard deviation
5.Index of maximum value
6.Index of minimum value
7.Number of values greater than the mean

```python
a = np.random.randint(1,101,(5,5))
print(a)
print("Maximum value:",np.max(a))
print("Minimum value:",np.min(a))
print("Mean:",np.mean(a))
print("Standard deviation:",np.std(a))
print("Index of maximum value:",np.argmax(a))
print("Index of minimum value:",np.argmin(a))
print("Number of values greater than the mean:",np.sum(a>np.mean(a)))
```
