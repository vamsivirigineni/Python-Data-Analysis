### 3. Operator Precedence

```python
# 1. Predict the output: arr=np.array([10,20,30]); arr+5*2
# Prediction: Multiplication (5*2=10) first, then addition (arr+10). [20, 30, 40]
arr = np.array([10, 20, 30])
arr + 5 * 2
```

```python
# 2. Predict the output: arr=np.array([5,10,15]); (arr+5)*2
# Prediction: Parentheses first (arr+5) = [10, 15, 20], then multiplication (*2) = [20, 30, 40]
arr = np.array([5, 10, 15])
(arr + 5) * 2
```

```python
# 3. Predict the output: arr=np.array([4,8,12]); arr/2+3
# Prediction: Division (arr/2) = [2, 4, 6], then addition (+3) = [5, 7, 9]
arr = np.array([4, 8, 12])
arr / 2 + 3
```

```python
# 4. Predict the output: arr=np.array([2,4,6]); arr*3-2
# Prediction: Multiplication (arr*3) = [6, 12, 18], then subtraction (-2) = [4, 10, 16]
arr = np.array([2, 4, 6])
arr * 3 - 2
```

```python
# 5. Predict the output: arr=np.array([5,10,15]); arr+10/2
# Prediction: Division (10/2=5) first, then addition (arr+5) = [10, 15, 20]
arr = np.array([5, 10, 15])
arr + 10 / 2
```

```python
# 5. Mini Project: A company stores the weekly sales of five employees:
sales = [25000, 32000, 28000, 35000, 30000]

# 1. Convert the list into a NumPy array.
sales_array = np.array(sales)
print("5. Mini Project:")
print("   1. Sales as NumPy array:", sales_array)

# 2. Find the total sales.
total_sales = np.sum(sales_array)
print("   2. Total sales:", total_sales)

# 3. Find the average sales.
average_sales = np.mean(sales_array)
print("   3. Average sales:", average_sales)

# 4. Add a £2,000 bonus to every employee.
bonus_sales = sales_array + 2000
print("   4. Sales with £2,000 bonus:", bonus_sales)

# 5. Find the highest and lowest sales.
highest_sales = np.max(sales_array)
lowest_sales = np.min(sales_array)
print("   5. Highest sales:", highest_sales)
print("      Lowest sales:", lowest_sales)

# 6. Display the updated sales array (with bonus).
print("   6. Updated sales array (with bonus):", bonus_sales)
```
