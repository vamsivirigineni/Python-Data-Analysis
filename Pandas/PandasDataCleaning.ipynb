Data Cleaning



```python
import pandas as pd
```

importing the data set

```python
df = pd.read_csv('/content/amazon_fires.csv',encoding = "latin1")
df
```

Step1:Renaming the column names:

```python
d = {
    'ano':'year',
    'estado':'state',
    'mes':'month',
    'numero':'number_of_fires',
    'encontro':'date'
}
df.rename(columns=d,inplace=True)
df
```

To check the duplicate values

```python
df[df.duplicated()]
int(df.duplicated().sum())
```

STEP2: droppingor removing the duplicate values

```python
df.drop_duplicates(inplace=True)
df
```

```python
df[df.duplicated()]
int(df.duplicated().sum())
```

Step3: removing the unwanted data from the column values

```python
df['number_of_fires']=df['number_of_fires'].str.strip(' Fires')
df
```

Step4: changing the datatype of the column

```python
df['number_of_fires']=df['number_of_fires'].astype(float)
df.info()
#df
```

STEP5: Changing the data in the columns

```python
d = {
    'Janeiro':'January',
    'Fevereiro':'February',
    'Março':'March',
    'Abril':'April',
    'Maio':'May',
    'Junho':'June',
    'Julho':'July',
    'Agosto':'August',
    'Setembro':'September',
    'Outubro':'October',
    'Novembro':'November',
    'Dezembro':'December'
}
df['month']=df['month'].map(d)
df
```

step6: checking the na(not available values):


```python
df.isna().any()
df.isna().sum()
```

step7: Fillig NA values

```python
df['number_of_fires']= df['number_of_fires'].fillna(0)
df
```

```python
int(df['number_of_fires'].isna().sum())
```

```python
df
```

step8: Rearranging the column order[link text](https://)

```python
new_order = [4,1,0,2,3]
df = df[df.columns[new_order]]
df
```

STEP9: drop unwanted columns

```python
#df=df.drop(['year','month'],axis=1)
#df
```

Step10: replacing the val in a column

```python
df['month']= df['month'].str.replace('January','Jan')
df
```
