```python
#Operations on Series
movies = pd.Series(['paradise','endcore','varanasi','raaka','dragon'],name="movies")
movies
```

```python
#1)Indexing on Series
print(movies[0])
print(movies[4])
print(movies[3])
```

```python
movies = pd.Series(['paradise','endcore','varanasi','raaka','dragon'],index=['nani','RDJ','BOB','AA','NTR'],name="movies")
movies
```

```python
#indexing on series based on label
print(movies['nani'])
print(movies['NTR'])
```

```python
#slicing on Series -> Indexing
print(movies[0:4])
```

```python
print(movies[0:5])
```

```python
print(movies[::2])
```

```python
print(movies[1::2])
```

```python
#slicing on series based on label
movies = pd.Series(['paradise','endcore','raaka','kaaka','salaar',"varanasi"],index=['nani','RDJ','AA','chiru','darling','bob'],name="movies")
movies
```

```python
print(movies['nani':'darling'])
```

```python
print(movies[:'AA'])
```

```python

#filtering on series:
#syntax: var[condition]
sal = pd.Series([15000,35000,20000,45000,50000,25000],name="salary")
sal
```

```python
print(sal[sal>30000])
```

```python
print(sal[sal<30000])
```

```python
print(sal[sal==50000])
```

```python
print(sal[(sal>30000)&(sal<50000)])
```

```python
print(sal[(sal>50000)|(sal<35000)])
```

```python
# not sal>30000
print(sal[sal<30000])
```

```python
print(sal[~(sal>30000)])
```

```python
nutrition={
    "items":["eggs","oats","sunflower seeds","dry fruits",
             "chicken","fish"],
    "clrs":[18,35,165,220,200,280],
    "prtn":[6,14,32,15,24,18],
    "fat":[3,2,7,15,10,12],
    "price":[7,250,120,700,250,140]

}
df=pd.DataFrame(nutrition)
df
```

```python
#filtering:
df[df["clrs"]<165]
df[df["prtn"]==18]
df[df["price"]>250]
df[(df["clrs"]<100)&(df["fat"]<10)]
df[(df["price"]>250)&(df["prtn"]>=15)]
df[(df["prtn"]<18)|(df["fat"]>15)]
```
