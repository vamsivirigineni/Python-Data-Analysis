```python
import pandas as pd
```

```python
#Series in pandas
sal = pd.Series([1000,1500,2000,2500,3000])
sal
```

```python
# Name attribute in Series
age = pd.Series([20,25,30,35,40],name = 'age')
age
```

```python
#Index in Series
age = pd.Series([20,21,22,23,24,25],index = ['sai','vamsi','sam','tarak','bob','bhAAi'],name="age")
age
```

```python
#Operations on Series
movies = pd.Series(['paradise','endcore','varanasi','raaka','dragon'],name="movies")
movies
```

```python
#1)Indexing on Series
print(movies[0])
print(movies[4])
print(movies[3])
```

```python
movies = pd.Series(['paradise','endcore','varanasi','raaka','dragon'],index=['nani','RDJ','BOB','AA','NTR'],name="movies")
movies
```

```python
#indexing on series based on label
print(movies['nani'])
print(movies['NTR'])
```

```python
#slicing on Series -> Indexing
print(movies[0:4])
```

```python
print(movies[0:5])
```

```python
print(movies[::2])
```

```python
print(movies[1::2])
```

```python
#slicing on series based on label
movies = pd.Series(['paradise','endcore','raaka','kaaka','salaar',"varanasi"],index=['nani','RDJ','AA','chiru','darling','bob'],name="movies")
movies
```

```python
print(movies['nani':'darling'])
```

```python
print(movies[:'AA'])
```
