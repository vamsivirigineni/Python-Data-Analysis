Extract the year, month number, month name, and day name from order_date into separate columns.

```python
df['year'] = df['order_date'].dt.year
df['month_number'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df
```

 Create a new column named price_category based on the following conditions:

Price greater than ₹20,000: "Expensive"
Price between ₹5,000 and ₹20,000 inclusive: "Medium"
Price below ₹5,000: "Budget"

```python
df.loc[df['price'] > 20000, 'price_category'] = 'Expensive'
df.loc[(df['price'] >= 5000) & (df['price'] <= 20000), 'price_category'] = 'Medium'
df.loc[df['price'] < 5000, 'price_category'] = 'Budget'
df
```


Create a new column named order_size based on the quantity ordered:

Quantity = 1: "Single"
Quantity = 2: "Small"
Quantity greater than 2: "Bulk"

```python
df.loc[df['quantity'] == 1, 'order_size'] = 'Single'
df.loc[df['quantity'] == 2, 'order_size'] = 'Small'
df.loc[df['quantity'] > 2, 'order_size'] = 'Bulk'
df
```
