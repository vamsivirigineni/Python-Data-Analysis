```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'order_id' : [1,2,3],
    'order_name' : ['bag','pen','laptop'],
    'order_price' : [1000,20,30000]
})
df1
```

```python
df2 = pd.DataFrame({
    'order_id' : [4,5,6],
    'order_name' : ['blanket','pillow','matress'],
    'order_price' : [100,200,15000]
})
df2
```

```python
data1 = {
  'customer_id': ['CUST501','CUST782','CUST501','CUST990'],
  'customer_name': ['john','BOB','Max','Anu'],
  'city': ['Banglore','Hyderabad','Banglore','Chennai'],
  'quantity' : [1,2,3,1],
  'product_id' : ['P101','P102','P103','P105']
}
order = pd.DataFrame(data1)
order
```

```python
data = {
    'product_id' : ['P101','P102','P103','P104'],
    'product_name' : ['Laptop','Mouse','Keyboard','Pen drive'],
    'Price': [50000, 500, 1500,500]
}
product = pd.DataFrame(data)
print(product)
```

1.CROSS JOIN

```python
pd.merge(order,product,how='cross')
```

2.INNER JOIN


```python
pd.merge(order,product,on='product_id',how='inner')
```

3.OUTER JOIN

```python
pd.merge(order,product,on='product_id',how='outer')
```

4.LEFT OUTER

```python
pd.merge(order,product,on='product_id',how='left')
```

5.RIGHT OUTER JOIN

```python
pd.merge(order,product,on='product_id',how='right')
```
