Q)Display the percentage distribution of bookings by Movie.

```python
data =df.groupby('Movie')['Ticket_ID'].count()
data
```

```python
plt.pie(data,labels=data.index,autopct='%0.1f%%', colors=["brown","gold","silver",'hotpink'],explode=[0,0.1,0,0])
plt.title('Percentage Distribution of Bookings by Movie')
plt.xlabel('Movie',color = 'lime')
plt.ylabel('Percentage',color='indigo')
plt.show()
```

Q)Represent the number of tickets sold for each Movie using a pie chart.

```python
x = df.groupby('Movie')['Tickets'].sum()
x
```

```python
plt.pie(x,labels=x.index,autopct='%0.1f%%',colors=['orange','blue','green','white'],wedgeprops={'edgecolor': 'black'})
plt.title('Number of Tickets Sold for Each Movie',color='red')
plt.gca().set_facecolor('green')
plt.gcf().set_facecolor('lightsalmon')
plt.xlabel('Movie',color = 'lime')
plt.ylabel('Tickets',color='indigo')
plt.show()
```

Q)Create a pie chart showing the top four movies by total tickets sold, grouping the remaining movies into an "Others" category.
