```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/scatterplot_study_hours_exam_score.csv')
df
```

```python
plt.figure(figsize=(10,5))
x = plt.bar(df['Study_Hours'], df['Exam_Score'], color='orange',width=0.4)
plt.bar_label(x)
plt.grid()
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.title('Scatter Plot of Study Hours vs Exam Score')
plt.show()
```

```python
df2= pd.read_csv('/content/piechart_monthly_expenses.csv')
df2
```

```python
df3 = pd.read_csv('/content/histogram_daily_commute_time.csv')
df3
```

```python
df4=pd.read_csv('/content/histogram_daily_commute_time.csv')
df4
```

```python
n,bins,p=plt.hist(df4['Commute_Minutes'],edgecolor="red",color='lightgreen',bins=10)
# plt.bar_label(p,)
plt.bar_label(p, labels=[int(x) for x in n])
# n=[1,3,3,8,...]--> Frequency of values
# bins = bar intervals (ex:0-30,30-40,....)
# p= considering the exact bin/bar to assign the label
plt.xlabel('Commute Minutes')
plt.ylabel('Frequency')
plt.title('Histogram of Commute Minutes')
plt.show()
```

```python
data = {
    "Day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday",
            "Saturday", "Sunday", "Monday", "Tuesday", "Wednesday",
            "Thursday", "Friday", "Saturday", "Sunday", "Monday"],

    "Temperature": [25, 28, 30, 27, 32, 35, 34, 29, 31, 36, 24, 33, 37, 26, 38],

    "Ice_Cream_Units_Sold": [120, 150, 180, 140, 220, 280, 260, 165,
                             200, 310, 100, 240, 340, 130, 360],

    "Customers": [85, 100, 125, 95, 160, 210, 195, 115, 145, 230,
                  75, 175, 250, 90, 270]}
```

```python
data1 = pd.DataFrame(data)
data1
```

```python
plt.figure(figsize=(14, 10))

# 1. Line Chart
plt.subplot(2, 2, 1)
plt.plot(data1['Day'],
         marker='o', mfc='orange', mec='black',
         markersize=8, color='green', linewidth=2,
         linestyle='solid', label='Temperature')
plt.xlabel('Day')
plt.ylabel('Temperature')
plt.title('Line Chart of Temperature vs Day')
plt.xticks(rotation=45)
plt.legend()

# 2. Bar Chart
plt.subplot(2, 2, 2)
plt.bar(data1['Day'], data1['Ice_Cream_Units_Sold'],
        color='orange', width=0.4)

plt.xlabel('Day')
plt.ylabel('Ice Cream Units Sold')
plt.title('Bar Chart of Ice Cream Units Sold vs Day')
plt.xticks(rotation=45)

# 3. Scatter Plot
plt.subplot(2, 2, 3)
plt.scatter(data1['Day'], data1['Customers'],
            color='black', s=50, alpha=0.5)

plt.xlabel('Day')
plt.ylabel('Customers')
plt.title('Scatter Plot of Customers vs Day')
plt.xticks(rotation=45)

# 4. Pie Chart
plt.subplot(2, 2, 4)
plt.pie(
    df2['Amount'],
    labels=df2['Expense_Category'],
    autopct='%0.1f%%',
    colors=['olive', 'blue', 'pink', 'lime',
            'orange', 'purple', 'cyan'],
    wedgeprops={'edgecolor': 'black'}
)
plt.xlabel('Expense Category')
plt.ylabel('Amount')
plt.title('Distribution of Monthly Expenses')

# Adjust spacing before displaying
plt.tight_layout()
plt.show()
```
