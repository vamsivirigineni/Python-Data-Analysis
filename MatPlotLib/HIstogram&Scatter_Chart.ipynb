Histogram:

DEF:
*Histogram is the graphical representaion of frequency distribution of the continuous quantitative data.

*histogram is mainly used to understand the distribution of the data based on one column.

*& here it's mandatory to mention numeric values on x-axis and by default.

*Histogram itself create count on y-axis.

*It groups numerical values to intervals called bins and show how many values fall into each interval.

*Histogram is helps us to understand how numerical values are distributed across different ranges

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Student_ID": list(range(1, 44)),
    "Marks": [
        25, 36, 38, 40, 42, 45, 47, 50, 52, 50,
        54, 58, 60, 70, 88, 93, 56, 62, 71, 79,
        84, 91, 47, 54, 66, 35, 82, 95, 59, 68,
        72, 80, 87, 50, 63, 76, 83, 92, 57, 65,
        70, 96, 98 ],
    "Section": [
        "A", "B", "A", "A", "B", "A", "B", "A", "B", "B",
        "A", "B", "A", "B", "A", "A", "B", "B", "A", "A",
        "B", "A", "B", "A", "B", "A", "B", "A", "A", "B",
        "B", "A", "A", "B", "A", "B", "A", "A", "B", "B",
        "A", "B", "A"]}
df = pd.DataFrame(data)
df
```

```python
plt.hist(df["Marks"],)
```

```python
plt.hist(df["Marks"],edgecolor="red",color='lightgreen')
plt.title("Histogram",color='hotpink')
plt.xlabel("Marks",color='indigo')
plt.ylabel("Frequency",color='Orange')
plt.show()
```

```python
n,bins,p=plt.hist(df["Marks"],edgecolor="red",color='lightgreen',bins=10)
# plt.bar_label(p,)
plt.bar_label(p, labels=[int(x) for x in n])
# n=[1,3,3,8,...]--> Frequency of values
# bins = bar intervals (ex:0-30,30-40,....)
# p= considering the exact bin/bar to assign the label
plt.title("Histogram",color='hotpink')
plt.xlabel("Marks",color='indigo')
plt.ylabel("Frequency",color='Orange')
plt.show()
```
