*1)Line Chart:

Def:

To know how values are changes over time, we can implement/use line Chart.



```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('/content/Website visitors')
df
```

```python
df = df.drop(['Unnamed: 0'],axis = 1)
df
```

```python
plt.plot(df['day'],df['mobile_visitors'],marker='o',mfc='orange',mec='black',markersize=10,color='green',linewidth=2,linestyle='solid',label='mobile_visitors')
plt.show()
```

```python
plt.plot(df['day'],df['mobile_visitors'],marker='o',mfc='orange',mec='black',markersize=10,color='green',linewidth=2,linestyle='solid',label='mobile_visitors')
plt.plot(df['day'],df['total_visitors'],marker='o',mfc='black',mec='white',markersize=10,color='blue',linewidth=2,linestyle='solid',label='totla_visitors')
plt.xlabel('day',color='hotpink')
plt.ylabel('no_of_visitors',color='indigo')
plt.title('website visitors',color='cyan')
plt.legend()
plt.show()

```

*2)Bar Chart:

def:

Bar Chart is used to compare the values across different categories.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bar1 = pd.read_csv('/content/Sales')
bar1
```

```python
plt.figure(figsize=(12,6))
x = plt.bar(bar1['Product'],bar1['Units Sold'],width=0.7,color='cyan')
plt.bar_label(x)
#plt.grid()
plt.xlabel('Product',color='hotpink')
plt.ylabel('Units Sold',color='brown')
plt.title('Sales',color='black')
plt.show()
```

```python
plt.figure(figsize=(12,6))
x = plt.bar(bar1['Product'],bar1['Revenue'],width=0.7,color='cyan')
plt.bar_label(x)
#plt.grid()
plt.xlabel('Product',color='hotpink')
plt.ylabel('Revenue',color='brown')
plt.title('Sales',color='black')
plt.show()
```

```python
bar2 = pd.read_csv('/content/Department vs employees')
bar2
```

```python
plt.figure(figsize=(6,5))
x = plt.bar(bar2['department'],bar2['employees'],width=0.4,color='indigo')
plt.bar_label(x)
plt.xlabel('department',color='navy')
plt.ylabel('employees',color='maroon')
plt.title('Department vs employees',color='black')
plt.show()
```

```python
bar3 = pd.read_csv('/content/Department vs salary')
bar3
```

```python
bar4 = pd.read_csv('/content/Course vs students')
bar4
```

```python
plt.figure(figsize=(6,5))
x = plt.bar(bar4['course'],bar4['students'],width=0.4,color='orange')
plt.bar_label(x,color='white')
plt.gca().set_facecolor('green')
plt.gcf().set_facecolor('lightsalmon')
plt.xlabel('course',color='navy')
plt.ylabel('students',color='blue')
plt.title('Course vs students',color='blue')
plt.show()
```

*3)Pie Chart:

def:

A pie chart is a circular statistical graph divided into slices (called wedges) to represent data proportions, where the area of each slice is proportional to the fraction or percentage of the category it represents relative to the whole.

```python
data = {
    "Order_ID": ["O101", "O102", "O103", "O104", "O105", "O106", "O107", "O108"],
    "Item": ["Pizza", "Burger", "Pasta", "Pizza", "Burger", "Pasta", "Pizza", "Burger"],
    "Category": ["Italian", "Fast Food", "Italian", "Italian", "Fast Food", "Italian", "Italian", "Fast Food"],
    "Quantity": [2, 3, 1, 4, 2, 5, 3, 2],
    "Revenue": [600, 450, 350, 1200, 300, 750, 900, 300]
}

df = pd.DataFrame(data)
df
```

Q)What percentage of revenue comes from each food item?

```python
data = df.groupby('Item')['Revenue'].sum()
data
```

```python
plt.pie(data,labels=data.index,autopct='%0.1f%%',colors=['brown','gold','silver'],explode=[0,0.1,0])
plt.title('Total_Revenue')
plt.show()
```

Q)Which category has largest proportion of revenue?

```python
cat=df.groupby('Category')['Revenue'].sum()
```

```python
plt.pie(cat,labels=cat.index,autopct="%0.1f%%",
        colors=["red","yellow"],
        explode=[0,0.1])
plt.axis('equal')
plt.title("total_revenue_by_each_catg",color="Coral")
plt.show()
```
