```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('/content/Website visitors')
df
```

```python
df = df.drop(['Unnamed: 0'],axis = 1)
df
```

```python
plt.plot(df['day'],df['mobile_visitors'],marker='o',mfc='orange',mec='black',markersize=10,color='green',linewidth=2,linestyle='solid',label='mobile_visitors')
plt.show()
```

```python
plt.plot(df['day'],df['mobile_visitors'],marker='o',mfc='orange',mec='black',markersize=10,color='green',linewidth=2,linestyle='solid',label='mobile_visitors')
plt.plot(df['day'],df['total_visitors'],marker='o',mfc='black',mec='white',markersize=10,color='blue',linewidth=2,linestyle='solid',label='totla_visitors')
plt.xlabel('day',color='hotpink')
plt.ylabel('no_of_visitors',color='indigo')
plt.title('website visitors',color='cyan')
plt.legend()
plt.show()

```

```python
df1 = pd.read_csv('/content/Time vs Temperature')


```

```python
df1
```

```python
plt.plot(df1['time'],df1['temperature'],marker='o',mfc='blue',mec='white',markersize=10,color='olive',linewidth=2,linestyle='dotted',label='temperature')
plt.xlabel('time',color='hotpink')
plt.ylabel('temperature',color='indigo')
plt.title('time vs temperature',color='#A104C3')
plt.legend()
plt.show()
```

```python
df2 = pd.read_csv('/content/Test vs Marks')
df2
```

```python
plt.plot(df2['test'],df2['marks'],marker='o',color='orange',mfc='blue',mec='white',linewidth=2,linestyle='dashdot',markersize=10,label='marks')
plt.xlabel('test',color='hotpink')
plt.ylabel('marks',color='indigo')
plt.title('test vs marks',color='green')
plt.legend()
plt.show()
```
